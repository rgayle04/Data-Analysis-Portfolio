<ul>Habitat: Typical environment where the animal is found</ul>
<ul>Predators: Natural organisms that prey on the animal.</ul>
<ul>Average Speed (km/h): The average speed range the animal can achieve in kilometers per hour.</ul>
<ul>Countries Found: regions where the animal is commonly found</ul>
<ul>Conservation Status: The conservation status of the animal as per relevant conservation organizations.</ul>
<ul>Family: Taxonomic family of the animal</ul>
<ul>Gestation Period (days): Range of days representing the pregnancy period of the animal</ul>
<ul>Top Speed (km/h): The maximum speed the animal can achieve in kilometers per hour.</ul>
<ul>Social Structure: Information about the social behavior or structure of the animal</ul>
<ul>Offspring per Birth: The typical number of offspring born per birth</ul>

<h2>Dataset Cleaning</h2>

```python
upto_columns= ['Height (cm)','Weight (kg)','Lifespan (years)','Offspring per Birth']
```

```python
for x in upto_columns:
    regex='Up.*'
    df[x][df[x].str.match(regex)]=df[x][df[x].str.match(regex)].str.split(' ',expand=True)[2]
```

```python
for x in ['Height (cm)', 'Weight (kg)', 'Lifespan (years)', 'Offspring per Birth', 'Average Speed (km/h)','Top Speed (km/h)', 'Gestation Period (days)']:
    ab=df[x][df[x].str.contains('-')]
    df[x][df[x].str.contains('-')]=ab.str.split('-').str.get(-1)
    
```

```python
df['Offspring per Birth'][df['Offspring per Birth'].str.endswith('(usually)')]=3
```

```python
for x in ['Countries Found', 'Habitat', 'Predators']:
    split_columns = df[x].str.split(', ', expand=True)
    split_columns.columns = [f'{x} {i+1}' for i in range(split_columns.shape[1])]

    df = pd.concat([df, split_columns], axis=1)
    df.drop(columns=[x], inplace=True )
    
```

```python
# Initialize an empty list to store the non-convertible values
non_convertible_values = []

# Iterate through the column and attempt to convert to float
for value in df['Offspring per Birth']:
    try:
        float(value)
    except ValueError:
        non_convertible_values.append(value)

# Create a DataFrame containing the non-convertible values
df.loc[df['Offspring per Birth'].isin(non_convertible_values), 'Offspring per Birth'] = df.loc[df['Offspring per Birth'].isin(non_convertible_values), 'Offspring per Birth'].str.split(" ",expand=True)[0]
```

<h3> The Tree Kangaroo </h3>
<img src='https://d1jyxxz9imt9yb.cloudfront.net/medialib/4061/image/s1300x1300/AdobeStock_576462084_441362_reduced.jpg' alt="Tree Kangaroo" >

<p>The Tree Kangaroo is a species of tree-based marsupials found in the rainforests of Papau New Guinea, Indonesia and Australia. There are 14 subspecies of tree kangaroo (My favorite is Huon/Matsie's Tree Kangaroo) with the majority of them being vulnerable, endangered or critically endangered.</p>
<p>
These are the following subspecies: Lumholtz’s tree kangaroo, Bennett’s tree kangaroo, Doria’s tree kangaroo, Lowlands tree kangaroo, Vogelkop tree kangaroo, Seri’s tree kangaroo, Grizzled tree kangaroo, Huon tree kangaroo, Goodfellow’s tree kangaroo, Dingiso tree kangaroo, Ifola tree kangaroo, Tenkile tree kangaroo, Wondiwoi tree kangaroo, Golden-mantled tree kangaroo.</p>

<p>We will not focus on a particular subspecies for this report, this will mainly be around the species of tree kangaroo in general.</p>
<p>Tree Kangaroos lead herbivorous diets primarily feeding on the foliage of New Guinea and Australia. The New Guinea tree-kangaroos tend to have a preference for figs, ferns, and ginger.(Martin & CSIRO Publishing, 2005, 39-40) The Australian tree-kangaroos seemed to have preference for epiphytic ferns, mature leaves, flowers of the Black Bean and Pepperberry Trees, and some vines. They also were surprisingly caught feeding on wild tobacco plants and the <i>Lantana camara</i> which is poisonous to most mammals.(Martin & CSIRO Publishing, 2005, 42)</p>

```python
df[df['Animal']=='Tree Kangaroo']
```

<p>According to <i>Tree-kangaroos of Australia and New Guinea</i> Tree Kangaroo's main issues regarding survival stem from over hunting and loss of habitat. The presence of hunters and european settlers had drastically lowered tree kangaroo populations in Australia, due not only to the humans themselves but their dogs. Their hunting dogs had managed to get loose sometimes and provide a newer predator to the environment to the tree kanagroos besides the already competitive dingos. Furthermore the dingoes and wild dogs would interbreed creating more feral canine predators for them to contend with. The text states, "...dingoes can rightly be regarded as a ‘natural’ predator of tree-kangaroos. However, the delicate equilibrium that once existed between them was upset by the arrival of European settlers and their soul mates, domesticated dogs. Many of these have become feral, interbred with and boosted the fecundity of the wild dog populations (dingoes only have one litter a year whereas domestic dogs can have two). Many other nominally domestic dogs live a substantially feral existence, roaming free at night and preying on wildlife. The net result is that tree-kangaroos now have more canine predators than ever before to contend with."(Martin & CSIRO Publishing, 2005, 136) Furthermore their deforestation of the areas not only stripped them of their homes but also provided less cover for the tree-kangaroos to have to avoid predators. Forcing them into the open grasslands where they would be easier targets for their canine predators. The text states, "...And because of European settlers’ widescale clearing of forest, tree-kangaroos often have to cross wide stretches of open country when moving between forest patches and, in such an exposed situation, they are easy meat for dogs." (Martin & CSIRO Publishing, 2005, 136).  </p>

<h3>The Pangolin</h3>
<img src = 'https://www.fauna-flora.org/wp-content/uploads/2023/06/AdobeStock_246102033_cropped-2-scaled.jpeg' alt = 'Pangolin'/>

<p>The Pangolin are mammals of the Manidae family, nicknamed the "scaly anteaters". Despite their nickname the two species animals are unrelated and most of the similarities are due to covergent evolution. Pangolin can be found around parts of Asia and Africa residing in burrows with 10 subspecies of them. In Asia there are the Indian pangolin, Philippine pangolin, Sunda pangolin, Chinese pangolin, Asian mysterious pangolin and the Indo-Burmese pangolin. In Africa there are the Black-bellied pangolin, White-bellied pangolin, Giant Ground pangolin, and Temminck's Ground pangolin. All of these subspecies of pangolin are listed as being either vulnerable, endangered, or critically endangered.</p>

<p>These pangolin are all observed as insectivores having a diet consisting of ants and termites similar to anteaters, contributing to its nickname, using its massive claws to break into the mounds and feed on them. The other part of its nickname being due to the massive scaly 'armor' surrounding it's body, consisting of epidermal scales, also leading to them being called 'walking pinecones' or 'walking artichokes'. The armor is used for defense from predators and the pangolin will also curl up into a ball to defend itself along with using its scaled up tail to also lash out against predators such as big cats(Lions & Tigers), chimpanzee, crocodiles, and python(ironic).</p>

```python
df[df['Animal']=='Pangolin']
```

<p>According to the WWF and <i>Pangolins: Science, Society and Conservation</i> the pangolins main threats to the conservation status is poaching, hunting and the illegal wildlife trade. Pangolins have been hunted and poached for several reasons, most notably their meat, skin, and scales. Pangolin meat is treated as a symbol of status for the wealthy, due to the price and rarity, with diners that serve it treating the entire process as a spectacle despite the animal's suffering. The scales of pangolin are also thought to have medicinal properties in traditional Chinese medicine. Despite being disproven due to their scales being made of the same material our hair and fingernails are made of, keratin, with no medinical properties discovered. The text states, "This was for the consumption of meat, favored in southern China, with assumed health benefits (see Chapter 14), and the use of scales in TCM. Such was the magnitude of this exploitation that it apparently led to the commercial extinction of pangolins in China by the mid-1990s. Wu et al. (2004) estimated that pangolin populations in China declined by up to 94% between the 1960s and early 2000s."  (Challender et al., 2019, 265) Furthermore pangolin have been trafficked as well for leather products, such as belts, boots and bags in the United States.</p>

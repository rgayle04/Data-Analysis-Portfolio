# <strong>Introduction</strong>


 <p style="text-indent: 25px;"> The field of psychology has tried to explain and understand personality through a variety of theories. The most popular and accepted theory to describe personality through the trait approach is the <strong>Big Five Traits</strong>, also known as the <strong>Five Factor Model (OCEAN)</strong>. The Big Five Model of Personality focuses on five main traits to describe personality:<strong> Openness, Conscientiousness, Extroversion, Agreeableness,</strong> and <strong>Neuroticism</strong>. 
</p>
<p style="text-indent: 25px;"/>    <strong>Openness(OPN)</strong> refers to how imaginative or intellectually curious an individual is. It measures how open they are to new ideas and exploring new experiences. Higher ratings of openness show being imaginative, open-minded, and curious, while lower measures indicate being closed-minded and preferring more traditional experiences. <strong>Conscientiousness(CSN)</strong> refers to how disciplined or hardworking someone may be. The trait measures how disciplined, reliable, hardworking, and efficient someone is. Higher measures of Conscientiousness show strong reliability and responsibility, while lower measures indicate impulsivity and overall disorganization. <strong>Extroversion(EXT)</strong> refers to the tendency someone has to seek out social interaction and react to social situations. Higher measures involve very high energy, risky, outgoing, and talkative tendencies, while lower measures are depicted as being less assertive, but more cautious and reserved. <strong>Agreeableness(AGR)</strong> refers to how friendly or cooperative an individual is. Scoring high in agreeableness indicates a strong sense of empathy and concern for the welfare of others, while lower scores show a more competitive nature and more selfish tendencies. <strong>Neuroticism(EST)</strong> describes how emotionally stable someone is. Higher scores of neuroticism correlate with even-temperedness and calm dispositions, however, lower scores are more vulnerable to negative emotions and stress. 
</p>
<p style="text-indent: 25px;">These traits, together at varying levels, make up the basis of people’s personalities. So, is it possible for these traits to allow us to predict the levels of another trait? Can we predict how conscientious someone is by how emotionally stable they are? Can we understand how emotionally stable someone is by their level of agreeableness? Can the OCEAN Model be reduced from five big traits to three or four? The data was gathered from the 50-question-long online survey by Open Psychometrics, with ten questions for each trait. However, some participants may find 50 questions too intimidating or even lie about some questions due to personal reasons, or wanting to ‘fit in’, or even misunderstanding the question. So, is it possible to reduce the questions asked in the survey from 10 questions in each category to around five or perhaps even two? Can we even guess the answers to questions from questions within the same category, or perhaps even different ones?  </p>

<p>

Recent studies have reported that heritability is not fixed, but instead decreases across the life span. However, findings are inconsistent and it is yet unclear whether these trends are because of a waning importance of heritable tendencies, attributable to cumulative experiential influences with age, or because of nonlinear patterns suggesting Gene × Environment interplay. We combined four twin samples (N = 7,026) from Croatia, Finland, Germany, and the United Kingdom, and we examined age trends in genetic and environmental variance in the six HEXACO personality traits: Honesty-Humility, Emotionality, Extraversion, Agreeableness, Conscientiousness, and Openness. The cross-national sample ranges in age from 14 to 90 years, allowing analyses of linear and nonlinear age differences in genetic and environmental components of trait variance, after controlling for gender and national differences. The amount of genetic variance in Extraversion, Agreeableness, and Openness followed a reversed U-shaped pattern across age, showed a declining trend for Honesty-Humility and Conscientiousness, and was stable for Emotionality. For most traits, findings provided evidence for an increasing relative importance of life experiences contributing to personality differences across the life span. The findings are discussed against the background of Gene × Environment transactions and interactions.

<p style="text-indent: 25px;">The data was then also normalized by shifting from a 1-5 scale for rating to a <b> -2 to 2 scale</b> for rating, and questions were classified as <i>“negative”</i> or <i>“positive”</i> in terms of how it affects the scoring of the question. Negative Questions are questions that test against a trait, and a higher rating for those questions will lower the overall score for that trait, while the Positive questions will test for the trait, and a higher rating for those questions will raise the overall score for that trait. For example in the Extroversion category a Negative question would be <i>"I don't talk a lot."</i>, while a Positive question may be <i>"I am the life of the party."</i> It is good to note as well that not all categories have the same number of negative and positive questions. The code that classifies questions as either negative or positive was made by <a href="https://www.kaggle.com/datasets/tunguz/big-five-personality-test/discussion/132061">Petar Luketina on Kaggle. </a> </p>

```python
#Code came from Petar Luketina on Kaggle
# positive questions adding to the trait.

pos_questions = [ 
    'EXT1','EXT3','EXT5','EXT7','EXT9',                       # 5
    'EST1','EST3','EST5','EST6','EST7','EST8','EST9','EST10', # 8
    'AGR2','AGR4','AGR6','AGR8','AGR9','AGR10',               # 6
    'CSN1','CSN3','CSN5','CSN7','CSN9','CSN10',               # 6
    'OPN1','OPN3','OPN5','OPN7','OPN8','OPN9','OPN10',        # 7
]

# negative (negating) questions subtracting from the trait.
neg_questions = [ 
    'EXT2','EXT4','EXT6','EXT8','EXT10', # 5
    'EST2','EST4',                       # 2
    'AGR1','AGR3','AGR5','AGR7',         # 4
    'CSN2','CSN4','CSN6','CSN8',         # 4
    'OPN2','OPN4','OPN6',                # 3
]

bigFiveDataset[pos_questions] = bigFiveDataset[pos_questions].replace({1:-2, 2:-1, 3:0, 4:1, 5:2})
bigFiveDataset[neg_questions] = bigFiveDataset[neg_questions].replace({1:2, 2:1, 3:0, 4:-1, 5:-2})

bigFiveDataset.head(6)
```

<h2 style="text-indent: 20px;" font-size=25><strong>Question Reduction Model:</strong></h2>

<p style="text-indent: 25px;">For this section we will be testing the capacity for the number questions to be reduced. In order to do this there were 3 functions built for each level of question testing, the baseline of 10 questions per trait (All Question Checker), then 5 questions(Quintuple Question Checker) and then 2 questions(Double Question Checker). To prevent bias in the Quintuple and Double Question Checkers random sampling was built into them to ensure that each question all had an equal chance of being the first or last attribute for testing. The Random Forest Algorithm was used throughout all of these methods due its high speed while also retaining high accuracies across the board.</p>

```python
def allQuestionChecker(attr, traitStatus):
    #accepts list of attrs corresponding to a specific Trait for training 
    bfD3= bigFiveDataset[[attr[0], attr[1],attr[2], attr[3],attr[4], attr[5],attr[6], attr[7],attr[8], attr[9], traitStatus]]#.head(50000)
    X = bfD3.drop(traitStatus, axis=1 )
    y = bfD3[traitStatus]
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state=42)
    clf=RandomForestClassifier(n_estimators=100, random_state=42)
    clf.fit(X_train,y_train)
    
    y_pred=clf.predict(X_test)
    
    accuracy = np.mean(cross_val_score(clf, X, y, cv=10, scoring='accuracy'))
        
    print(traitStatus,'Accuracy:')
    print(accuracy)
    print() 
```

```python
import random
def quintQuestionChecker(attr, traitStatus):
    lst=[]
    while len(lst)< 10:
        available = [num for num in range(10) if num not in lst]

        if len(available) < 5:
            break
            
        n,m, p, l, o = random.sample(available,5)
        lst.extend([n,m,p, l, o])
        bfD3= bigFiveDataset[[attr[n], attr[m], attr[p],attr[l],attr[o], traitStatus]]#.head(50000)
        X = bfD3.drop(traitStatus, axis=1 )
        y = bfD3[traitStatus]
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state=42)
        clf=RandomForestClassifier(n_estimators=100, random_state=42)
        clf.fit(X_train,y_train)
        y_pred=clf.predict(X_test)
        accuracy = np.mean(cross_val_score(clf, X, y, cv=10, scoring='accuracy'))
            
        print(attr[n], attr[m], attr[p], attr[l], '&' ,attr[o],'Accuracy:')
        print(accuracy)
        print()
```

```python
def doubQuestionChecker(attr, traitStatus):
    lst=[]
    finalAccuracies=[]
    while len(lst)< 10: 
        available = [num for num in range(10) if num not in lst]

        if len(available) < 2:
            break
            
        n,m = random.sample(available,2)
        lst.extend([n,m])
        bfD3= bigFiveDataset[[attr[n], attr[m], traitStatus]]#.head(50000)
        X = bfD3.drop(traitStatus, axis=1 )
        y = bfD3[traitStatus]
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state=42)
        clf=RandomForestClassifier(n_estimators=100, random_state=42)
        clf.fit(X_train,y_train)
        
        # prediction on test set
        y_pred=clf.predict(X_test)
        
        # Model Accuracy, how often is the classifier correct?
        accuracy = np.mean(cross_val_score(clf, X, y, cv=10, scoring='accuracy'))
        finalAccuracies.append(accuracy)
        print(attr[n], '&', attr[m],'Accuracy:')
        print(accuracy)
        print()
    
```

<h2 style="text-indent: 20px;" font-size=25><b>Question to Question Prediction Model:</b></h2>

<p style="text-indent: 25px;">To perform this question to question guessing another function was built using the same format as the other question checker functions, the Triple Question Check which will take an array of nine questions/attributes and a targetted question and then randomly select three questions at a time and try to predict the answer to the target question. Another one was made again with the questToquest function which will try predicting the answers to a question based off a single question.</p>

```python
def tripQuestionChecker(attr, traitStatus):
    lst=[]
    while len(lst)< 9:
        available = [num for num in range(9) if num not in lst]

        if len(available) < 3:
            break
            
        n,m, p = random.sample(available,3)
        lst.extend([n,m,p])
        bfD3= bigFiveDataset[[attr[n], attr[m], attr[p], traitStatus]]#.head(50000)
        X = bfD3.drop(traitStatus, axis=1 )
        y = bfD3[traitStatus]
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state=42)
        clf=RandomForestClassifier(n_estimators=100, random_state=42)
        clf.fit(X_train,y_train)
        y_pred=clf.predict(X_test)
        accuracy = np.mean(cross_val_score(clf, X, y, cv=10, scoring='accuracy'))
            
        print(attr[n], attr[m],'&', attr[p],'Accuracy:')
        print(accuracy)
        print()
```

The Big Five Model empahsizes that these big five traits are core traits that build up the crux of people's personalities these traits work in tandem to make us who we are at various levels and each may be separately influence by a variety of factors, either genetic inherited features or environmental factors that influenced the way these traits are expressed. These traits are all very independent of one another. They are each the building blocks of a person's personality the same way that red, green, and blue make up the rgb color system. To try and assume that we could accurately guess the how Neurotic someone is based on how Conscientious they are is essentially the same as trying to guess the amount of red is in a picture based off the amount of green. It just cannot work that way. Previous Models of the Trait Approach to personality have been used before where it had been reduced to 3 core traits such as the PEN model. However, it was superceded by the Big Five Model due to greater support in empirical research and broader way of explaining personality in terms of dimensions. </p>
